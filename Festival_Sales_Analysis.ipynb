```python

sns.set(rc={'figure.figsize': (7,5)})

ax = sns.countplot(data=df, x='Marital_Status', palette='Set2')

for bars in ax.containers:
    ax.bar_label(bars)

plt.title("Count of Customers by Marital Status")
plt.show()
```

```python
# Group sales by Marital Status and Gender
sales_state = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

# Set figure size
sns.set(rc={'figure.figsize': (8,5)})

# Create grouped bar chart
sns.barplot(data=sales_state, x='Marital_Status', y='Amount', hue='Gender')
```

#### From the above graph we can see that most of the buyers are married(women) and they have high purchasing power 

# Occupation

```python
# Set figure size
sns.set(rc={'figure.figsize': (20,5)})

# Create count plot
ax = sns.countplot(data=df, x='Occupation')

# Add labels on top of bars
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Group sales by Occupation
sales_state = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

# Set figure size
sns.set(rc={'figure.figsize': (20,5)})

# Plot bar chart
sns.barplot(data=sales_state, x='Occupation', y='Amount')
```

#### From the above graph we can see that most of the buyers are working in IT, Aviation and Healthcare sector 

# Product Category

```python

# Set figure size
sns.set(rc={'figure.figsize': (20,5)})

# Create count plot with colorful bars using palette
ax = sns.countplot(data=df, x='Product_Category', palette='Set2')

# Add numbers on top of bars
for bars in ax.containers:
    ax.bar_label(bars)

# Show plot
plt.show()
```

```python
# Group by Product_Category and sum Amount
sales_state = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

# Set figure size
sns.set(rc={'figure.figsize': (20, 5)})

# Create bar plot
sns.barplot(data=sales_state, x='Product_Category', y='Amount')

# Optional: show plot
plt.show()
```

#### From the above graph we can see that most of the sold products are from food, Clothing and Electronics
